```python
def square(x):
    return x * x

numbers = [1, 2, 3, 4]

result = list(map(square, numbers))
print(result)
```

```python
def is_even(x):
    return x % 2 == 0

numbers = [1, 2, 3, 4, 5, 6]

result = list(filter(is_even, numbers))
print(result)
```

```python
## Inheritence

class Parent:
    def display(self):
        return "Hello AI"
class Child(Parent):
    def show():
        return "World"

c1=Child()
print(c1.display())
```

```python
class GrandA:
    def grd(self):
        return "Grad Father is calling"
class Parent(GrandA):
    pass
class Child(Parent):
    def show(self):
        return"This is child"
h1= Child()
print(h1.grd())
```

```python
## Multi inhertance 

class ClassA:
    def show(self):
        return"Class A"
class ClassB:
    def display(self):
        return"Class B"
class ClassC(ClassA,ClassB):
    pass

sr= ClassC()
print(sr.display())
print(sr.show())
b= ClassA()
print(b.show())
```

```python
## Hierarchical

class Animal:
    def speak(self):
        return"Animal "

class Dog(Animal):
    def bark(self):
        return"Dog "

class Cat(Animal):
    def meow(self):
        return"Cat "

d = Dog()
print(f"{d.speak()}{d.bark()}")
c = Cat()
print(f"{c.speak()}{c.meow()}")
```

```python
## Hybrid

class A:
    def showA(self):
        return"Class A"

class B(A):
    def showB(self):
        return"Class B"

class C(A):
    def showC(self):
        return"Class C"

class D(B, C):
    def showD(self):
        return"Class D"

obj = D()
print(f"{obj.showA()}\n{obj.showB()}\n{obj.showC()}\n{obj.showD()}")



```
