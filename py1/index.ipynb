```python
## Multi inhertance 

class ClassA:
    def show(self):
        return"Class A"
class ClassB:
    def display(self):
        return"Class B"
class ClassC(ClassA,ClassB):
    pass

sr= ClassC()
print(sr.display())
print(sr.show())
b= ClassA()
print(b.show())
```

```python
## Hierarchical

class Animal:
    def speak(self):
        return"Animal "

class Dog(Animal):
    def bark(self):
        return"Dog "

class Cat(Animal):
    def meow(self):
        return"Cat "

d = Dog()
print(f"{d.speak()}{d.bark()}")
c = Cat()
print(f"{c.speak()}{c.meow()}")
```

```python
## Hybrid

class A:
    def showA(self):
        return"Class A"

class B(A):
    def showB(self):
        return"Class B"

class C(A):
    def showC(self):
        return"Class C"

class D(B, C):
    def showD(self):
        return"Class D"

obj = D()
print(f"{obj.showA()}\n{obj.showB()}\n{obj.showC()}\n{obj.showD()}")



```

```python
# Encapsolution
# Protected

class Employee:
    def __init__(self,name,age):
        self.name = name
        self._age= age

class SubEmployee(Employee):
    def show(self):
        return self._age

p1= SubEmployee("Ali",20)
print(p1.show())
        
```

```python
## Bubble Sorting

numbers = [5, 3, 8, 4]

# Bubble sort
for i in range(len(numbers)):
    for j in range(len(numbers) - 1):
        if numbers[j] > numbers[j + 1]:
            # swap
            numbers[j], numbers[j + 1] = numbers[j + 1], numbers[j]

print(numbers)
```

```python
## Binary Sorting 

numbers = [1, 3, 1, 0, 1, 4]

# sort the list
numbers.sort()

print(numbers)
```

```python
## Linear Search

numbers = [5, 3, 8, 4]
target = 4

for i in range(len(numbers)):
    if numbers[i] == target:
        print("Found at index", i)
        break
```

```python
## Recursion

def countdown(n):
    if n == 0:   # base case (stop)
        print("Done")
    else:
        print(n)
        countdown(n - 1)   # function calls itself

countdown(3)
```

```python
## Factorial Mathmatical 

def factorial(n):
    if n == 1:   # base case
        return 1
    else:
        return n * factorial(n - 1)

print(factorial(5))
```

```python
## Stack

stack = []

# Push (add)
stack.append(10)
stack.append(20)
stack.append(30)

print(stack)   # [10, 20, 30]

# Pop (remove)
stack.pop()

print(stack)   # [10, 20]

# Peek (top element)
print(stack[-1])   # 20
```

```python
## Queue

from collections import deque

# Step 1: Create a queue
queue = deque()

# Step 2: Add elements to the queue (enqueue)
queue.append("apple")
queue.append("banana")
queue.append("cherry")

print("Queue after adding items:", queue)

# Step 3: Remove elements from the queue (dequeue)
first_item = queue.popleft()  # removes "apple"
print("Removed item:", first_item)
print("Queue now:", queue)

second_item = queue.popleft()  # removes "banana"
print("Removed item:", second_item)
print("Queue now:", queue)
```
